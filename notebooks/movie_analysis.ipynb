```python
import sqlite3
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

db_path = Path("../database/movies.db") if Path("../database/movies.db").exists() else Path("database/movies.db")
connection = sqlite3.connect(db_path)
df = pd.read_sql_query("SELECT * FROM movies", connection)
connection.close()

print("✅ Data loaded successfully")
df.head()
```

```python
print(df.info())
print(df.describe())

avg_rating_genre = df.groupby("genre")["rating"].mean()
print("\nAverage Rating per Genre:")
print(avg_rating_genre)

avg_revenue_year = df.groupby("release_year")["revenue"].mean()
print("\nAverage Revenue per Year:")
print(avg_revenue_year)

df["decade"] = (df["release_year"] // 10) * 10
movies_per_decade = df.groupby("decade").size()
print("\nMovies per Decade:")
print(movies_per_decade)

correlation = df[["rating", "revenue"]].corr()
print("\nCorrelation between Rating and Revenue:")
print(correlation)

top_movies_decade = df.groupby("decade", group_keys=False).apply(lambda x: x.nlargest(5, "rating"))
print("\nTop 5 Movies by Rating in Each Decade:")
print(top_movies_decade[["title", "rating"]])
```

```python
plt.figure(figsize=(8, 6))
sns.histplot(df['rating'], bins=20, kde=True, color="skyblue")
plt.title("Distribution of Movie Ratings")
plt.xlabel("Rating")
plt.ylabel("Number of Movies")
plt.grid(True)
plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 6))
sns.barplot(x=avg_rating_genre.index, y=avg_rating_genre.values, hue=avg_rating_genre.index, palette="viridis", legend=False)
plt.title("Average Rating by Genre")
plt.xlabel("Genre")
plt.ylabel("Average Rating")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 6))
sns.lineplot(x=avg_revenue_year.index, y=avg_revenue_year.values, marker="o", color="red")
plt.title("Average Revenue Trend Over Years")
plt.xlabel("Release Year")
plt.ylabel("Average Revenue ($)")
plt.grid(True)
plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 6))
sns.scatterplot(x="rating", y="revenue", data=df, hue="genre", palette="Set2")
plt.title("Rating vs Revenue")
plt.xlabel("Rating")
plt.ylabel("Revenue ($)")
plt.grid(True)
plt.tight_layout()
plt.show()

plt.figure(figsize=(6, 4))
sns.heatmap(df[["rating", "revenue", "release_year"]].corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```
